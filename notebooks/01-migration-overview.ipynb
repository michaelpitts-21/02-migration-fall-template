# Mapping migration

### An introduction to vector data operations

The Veery thrush (*Catharus fuscescens*) migrates each year between nesting sites in North America and South America. Veeries are small birds, and encountering a hurricane during their migration would be disasterous. However, Ornithologist Christopher Heckscher has found in tracking their migration that their breeding patterns and migration timing can predicting the severity of hurricane season as accurately as meterological models [(Heckscher 2018)](https://doi.org/10.1038/s41598-018-28302-3)!

### Reflect and Respond #1: Reflect on what you know about migration. You could consider:
1. What are some reasons that animals might migrate?
2. How might climate change affect animal migrations?
3. Do you notice any animal migrations where you live?

**Your answer here:**

1) Animals migrate for a few reasons, including weather, breeding, and food sources. Some animals are conditioned better for certain weather, which typically corresponds with available food. Here in Colorado, Bighorn Sheep are a great example of this. During the summer, they are at high altitudes around and above treeline because there is food and relatively low competition. During the cooler months, they move down to lower elevations because high altitudes no longer have food sources. Breeding migrations are typically done to areas where natural predators are less prevalent, in order to increase the rate of survival in offspring.


2) Climate change affects animal migrations in a variety of ways. The main issue that stands out to me is the disruption of timing climate change is beginning to cause. This can cause animals to show up to their traditional breeding grounds at the wrong time. This wrong time is predicated on two factors: either their food sources peaked earlier than normal or the animal showed up before the food source peaked. This can lead to dramatic effects on local ecosystems as the balance is thrown off completely compared to normal due to irregular habits. Additionally, sometimes migrations can overlap that did not traditionally overlap, leading to a depletion of local ecosystem resources.


3) In and around Boulder, I did not notice any animal migrations until this summer. I was out fishing at Coot Lake just north of Boulder Reservoir and noticed many birds in the water that appeared to be Pelicans. I figured I was seeing things, until I looked it up and found out that American White Pelicans come to the area to breed each spring/summer. To make this even more interesting, White Pelicans did not stop to breed in Colorado until the Front Range became developed. Prior to development, the Front Range lacked the water bodies necessary for them to breed, as they typically traveled further north to find natural bodies of water to breed. Their populations have experienced severe decline in the past, and ironically development along the Front Range gave them more opportunity to get their numbers to rebound.
